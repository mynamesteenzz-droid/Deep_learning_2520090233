```python
import json

annotation_path = "/content/coco_demo/annotations/instances_val2017.json"

with open(annotation_path, "r") as f:
    coco_data = json.load(f)

print("COCO JSON loaded successfully!")

print("Images:", len(coco_data["images"]))
print("Annotations:", len(coco_data["annotations"]))
print("Categories:", len(coco_data["categories"]))
```

```python
print("COCO CATEGORIES:\n")

for category in coco_data["categories"]:
    print(
        "Category ID:", category["id"],
        "| Class Label:", category["name"]
    )
```

```python
print("IMAGE RECORDS:\n")

for img in selected_images:
    print(img)
    print("-" * 60)
```

```python
selected_ids = {img["id"] for img in selected_images}

selected_annotations = [
    ann for ann in coco_data["annotations"]
    if ann["image_id"] in selected_ids
]

print("Total selected annotations:", len(selected_annotations))

for ann in selected_annotations[:10]:
    print(ann)
```

```python
category_map = {
    cat["id"]: cat["name"]
    for cat in coco_data["categories"]
}

for ann in selected_annotations[:10]:
    print(
        "Annotation ID:", ann["id"],
        "| Image ID:", ann["image_id"],
        "| Category:", category_map[ann["category_id"]],
        "| BBox:", ann["bbox"]
    )
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
import os

for img_info in selected_images:

    image_path = os.path.join(
        image_dir,
        img_info["file_name"]
    )

    image = Image.open(image_path).convert("RGB")

    fig, ax = plt.subplots(figsize=(10, 7))
    ax.imshow(image)

    # Find annotations belonging to this image
    anns = [
        ann for ann in selected_annotations
        if ann["image_id"] == img_info["id"]
    ]

    for ann in anns:

        x, y, w, h = ann["bbox"]

        # Draw bounding box
        rect = patches.Rectangle(
            (x, y),
            w,
            h,
            linewidth=2,
            edgecolor="red",
            facecolor="none"
        )

        ax.add_patch(rect)

        # Get class label
        label = category_map[ann["category_id"]]

        # Display class label
        ax.text(
            x, y - 5,
            label,
            color="white",
            fontsize=10,
            bbox=dict(facecolor="red", alpha=0.7)
        )

    ax.set_title(
        f"Image ID: {img_info['id']} | Objects: {len(anns)}"
    )

    ax.axis("off")
    plt.show()
```

```python
print("========== COCO DATASET SUMMARY ==========")

print("Total images in JSON:", len(coco_data["images"]))
print("Total annotations:", len(coco_data["annotations"]))
print("Total categories:", len(coco_data["categories"]))

print("\n========== SELECTED IMAGES ==========")

for img in selected_images:
    count = sum(
        1 for ann in selected_annotations
        if ann["image_id"] == img["id"]
    )

    print(
        "Image ID:", img["id"],
        "| Filename:", img["file_name"],
        "| Objects:", count
    )

print("\n========== PRACTICAL COMPLETED ==========")
```

```python
import matplotlib.pyplot as plt

image_names = []
object_counts = []

for img in selected_images:
    count = sum(
        1 for ann in selected_annotations
        if ann["image_id"] == img["id"]
    )

    image_names.append(str(img["id"]))
    object_counts.append(count)

plt.figure(figsize=(10, 5))
plt.bar(image_names, object_counts)

plt.title("Number of Objects in Each Image")
plt.xlabel("Image ID")
plt.ylabel("Number of Objects")
plt.xticks(rotation=45)
plt.show()
```

```python
from collections import Counter

category_counts = Counter(
    ann["category_id"] for ann in selected_annotations
)

labels = [
    category_map[cid] for cid in category_counts
]

values = list(category_counts.values())

plt.figure(figsize=(12, 6))
plt.bar(labels, values)

plt.title("Object Category Distribution")
plt.xlabel("Object Categories")
plt.ylabel("Number of Objects")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
bbox_areas = [
    ann["bbox"][2] * ann["bbox"][3]
    for ann in selected_annotations
]

plt.figure(figsize=(9, 5))
plt.hist(bbox_areas, bins=10)

plt.title("Bounding Box Area Distribution")
plt.xlabel("Bounding Box Area (pixels squared)")
plt.ylabel("Frequency")
plt.show()
```
